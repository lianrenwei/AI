```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Conv2D, MaxPooling2D, Flatten
model=Sequential()
model.add(Conv2D(filters=16, kernel_size=(5, 5), input_shape=(28, 28, 1), activation='relu'))  #建立卷積層
model.add(MaxPooling2D(pool_size=(2, 2)))  #建立最大池化層
model.add(Flatten())  #建立展平層
model.add(Dense(10, activation='softmax'))  #建立輸出層
model.summary()  #顯示模型架構
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['accuracy'])   #編譯模型

mnist=keras.datasets.mnist   #載入手寫數字資料集
(train_data, train_label), (test_data, test_label) = mnist.load_data()
test_label_original=test_label  #保留原本測試標籤
train_label = keras.utils.to_categorical(train_label)
test_label = keras.utils.to_categorical(test_label)

train_history=model.fit(train_data,train_label,epochs=10,batch_size=512)

score=model.evaluate(test_data, test_label)
print(score[1])


import matplotlib.pyplot as plt   #匯入繪圖套件
import numpy as np

def show_train_history(train_history, train, validation):  #定義訓練歷史紀錄繪圖函數
    plt.plot(train_history.history[train])
    plt.plot(train_history.history[validation])
    plt.title('Train History')
    plt.ylabel('Train')
    plt.xlabel('Epoch')
    plt.legend(['train', 'validation'], loc='center right')
    plt.show()

show_train_history(train_history, 'accuracy', 'accuracy')   #繪圖模型訓練準確率變化圖
show_train_history(train_history, 'loss', 'loss')  #繪圖模型訓練損失值變化圖


import pandas as pd
prediction = model.predict(test_data)
prediction_label=np.argmax(prediction,axis=1)
print(test_label.shape)  #印出測試標籤的形狀
pd.crosstab(test_label_original, prediction_label, rownames=['label'], colnames=['predict'])  #建立交叉分析表
```
