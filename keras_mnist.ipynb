```python
from tensorflow import keras
from keras.models import Sequential  #匯入模型
from keras.layers import Dense, Flatten  #匯入Dense全連階層 Flatten階層



mnist=keras.datasets.mnist
(train_data, train_label), (test_data, test_label) = mnist.load_data()   #資料分成訓練資料和測試資料

print(type(train_data))
print(train_data.shape)
print(type(train_label))
print(train_label.shape)
print(test_data.shape)
print(test_label.shape)

print(train_label[0])
print(train_data[0])

import matplotlib.pyplot as plt  #匯入繪圖套件
plt.imshow(train_data[0], cmap='binary')
plt.show()

model=Sequential()  #建立 神經網路模型
model.add(Flatten()) #建立 展平層
model.add(Dense(32, activation='relu', input_dim=784))  #建立第一層隱藏層
model.add(Dense(10, activation='sigmoid'))  #建立輸出層
model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy', metrics=['accuracy'])  #編譯神經網路模型

model.fit(train_data,train_label,epochs=20,batch_size=512)  #訓練模型

score=model.evaluate(test_data, test_label)  #測試資料評估訓練完成的模型
print(score[1])
```
