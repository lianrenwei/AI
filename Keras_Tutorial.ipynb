```python
from tensorflow import keras
from keras.models import Sequential  #匯入Sequential模型 依照順序建立神經網路
from keras.layers import Dense       #匯入Dense全連階層

import numpy as np     #匯入NumPy  處理數值 鎮列
data = np.random.random((1000, 10))
labels = np.random.randint(2, size=(1000, 1))
print(np.shape(data))
print(data[0])
print(np.shape(labels))
print(labels[0])

model = Sequential()   #建立Sequential 神經網路模型
model.add(Dense(32, activation='relu', input_dim=10))   #建立第一層(隱藏層)
model.add(Dense(1, activation='sigmoid'))               #建立第二層(輸出層)
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])  #編譯模型

model.fit(data, labels, epochs=10, batch_size=32)  #訓練模型

score=model.evaluate(data, labels)  #評估訓練完成模型
print(score[1])
```
